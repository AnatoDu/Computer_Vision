Функция ниже — очень простое правило для красного. До запуска определите его ответы. Найдите пропуск тёмно-красного и объясните, почему красная поверхность ещё не означает красный сигнал светофора.

```python
samples = [("ярко-красный", (240, 20, 20)), ("тёмно-красный", (100, 5, 5)), ("белый", (250, 250, 250)), ("жёлтый", (240, 240, 20))]
def is_red(rgb):
    r, g, b = rgb
    return r > 150 and r > 2 * g and r > 2 * b
for name, rgb in samples:
    print(name, is_red(rgb))
# Ваш ход: измените правило, запишите прежний и новый результаты.
```

## 6. Цвет и яркость — 20 минут

Стандартная библиотека colorsys позволяет увидеть идею HSV без OpenCV. Сравните ярко-красный и тёмно-красный: что меняется сильнее, H или V? Здесь H/S/V лежат в 0..1; в OpenCV uint8 шкалы другие. Белый и серый почти не имеют насыщенности: один H не решает задачу.

```python
import colorsys
for name, rgb in samples:
    hsv = colorsys.rgb_to_hsv(*(channel / 255 for channel in rgb))
    print(name, tuple(round(value, 3) for value in hsv))
# Ваш ход: добавьте серый пиксель и объясните S.
```
