### Негатив: цикл против NumPy

Сделаем негатив: каждое значение `v` заменим на `255 - v`. Сначала — «в лоб», циклом по всем пикселям и каналам:

```python
import time

start = time.perf_counter()
negative_loop = img.copy()
height, width, channels = img.shape
for y in range(height):
    for x in range(width):
        for c in range(channels):
            negative_loop[y, x, c] = 255 - img[y, x, c]
loop_seconds = time.perf_counter() - start

print(f'Цикл: {loop_seconds:.2f} с')
```

А теперь так, как принято в NumPy, — одной операцией над всем массивом:

```python
start = time.perf_counter()
negative_numpy = 255 - img
numpy_seconds = time.perf_counter() - start

print(f'NumPy: {numpy_seconds * 1000:.2f} мс — быстрее в {loop_seconds / numpy_seconds:.0f} раз')
print('Результаты совпадают:', np.array_equal(negative_loop, negative_numpy))
show(img, negative_numpy, titles=['оригинал', 'негатив'])
```

Откуда такая разница? NumPy делает то же самое в скомпилированном коде на C: идёт по непрерывному куску памяти и обрабатывает сразу несколько чисел за одну команду процессора (векторные инструкции — привет «Архитектуре компьютера»). А Python на каждом шаге цикла заново выясняет, что за объект перед ним и что с ним делать.

Для зрения это вопрос жизни и смерти. При видео 30 кадров в секунду на один кадр есть 33 миллисекунды — цикл не успел бы обработать даже нашу маленькую картинку.

**Забавный факт.** Грейс Хоппер раздавала на лекциях отрезки провода длиной около 30 сантиметров: столько свет пролетает за одну наносекунду. Так она объясняла, почему в быстром компьютере сигналам нельзя далеко бегать. Нам до наносекунд далеко, но правило то же: **вместо Python-циклов по пикселям предпочитаем операции над массивами**.

По умолчанию `cv2.resize` использует билинейную интерполяцию, а при сильном уменьшении лучше работает `interpolation=cv2.INTER_AREA`: она усредняет соседние пиксели и не даёт ряби.
